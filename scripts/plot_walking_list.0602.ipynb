```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
walking_list_file = Path('/astrum/home/hpchzy/code/TacVar/scripts/walklists/detecting_expr1_fsize_Normal_n20_tbase1000.csv')
df = pd.read_csv(walking_list_file, header=None, names=["ns"])
df[:10]
```

```python
fig, ax = plt.subplots()

x = np.sort(df["ns"].dropna().to_numpy())
y = np.arange(1, len(x) + 1) / len(x)

x_unique, idx = np.unique(x, return_index=True)
y_unique = y[idx]

xs = np.linspace(x_unique.min(), x_unique.max(), 500)
ys = np.interp(xs, x_unique, y_unique)

ax.plot(xs, ys, label="Interpolated ECDF")

ax.set_xlabel("ns")
ax.set_ylabel("CDF")
ax.grid(True, alpha=0.3)
ax.legend()
```
